### Create combined 2 row figure

```python
# ── Zero Annual Amplitude (ZAA) estimation ────────────────────────────────────
# Fits an exponential amplitude decay A(z) = A0 * exp(-z/d) to the measured
# seasonal amplitudes at each sensor depth. ZAA is reported only where the
# deepest sensor already has amplitude < 0.3°C (otherwise extrapolation is
# too uncertain). ZAA is defined where fitted amplitude < 0.05°C.

from scipy.optimize import curve_fit

_AMP_THRESHOLD = 0.3   # °C — skip ZAA if deepest sensor >= this
_ZAA_SIGNAL    = 0.05  # °C — amplitude level defining ZAA

_files = {
    "AH": [temp_path + 'AH1TT_20240821_20250916.csv',
           temp_path + 'AH2TT_20240821_20250916_spliced.csv'],
    "CJ": [temp_path + 'CJ1TT_20240809_20250808_spliced.csv',
           temp_path + 'CJ2TT_20240809_20250808_spliced.csv'],
    "HS": [temp_path + 'HS1TT_20240808_20250927_spliced.csv',
           temp_path + 'HS2TT_20240808_20250927_spliced.csv'],
    "SR": [temp_path + 'SR1TT_20240806_20250724_spliced.csv',
           temp_path + 'SR2TT_20240806_20250724_spliced.csv'],
    "GT": [temp_path + 'GT1TT_20240807_20250723_spliced.csv'],
    "CV": [temp_path + 'CT1TT_20240828_20250905.csv',
           temp_path + 'CT2TT_20240828_20250905_spliced.csv'],
}
# current depths per glacier [BH1_white, BH1_black, BH2_white, BH2_black]
_depths = {
    "AH": [9.57, 14.57, 7.81, 12.81],
    "CJ": [4.1,  9.1,  5.48, 10.48],
    "HS": [8.3,  13.3,  7.5,  12.5],
    "SR": [6.57, 11.82, 8.93, 13.38],
    "GT": [1.43, 6.43],
    "CV": [1.26, 6.26,  3.09,  8.09],
}

def _sensor_amplitudes(code):
    from src.thermistor_processing import ThermistorData
    pairs = []
    depth_list = _depths[code]
    d_idx = 0
    for fpath in _files[code]:
        td = ThermistorData(fpath, delimiter=',')
        df = td.get_ntc_data().sort_values('TIME')
        t0, t1 = df['TIME'].min(), df['TIME'].max()
        if (t1 - t0).days >= 330:
            t0 = t0 + pd.Timedelta(days=14)   # skip equilibration
        df = df[(df['TIME'] >= t0) & (df['TIME'] <= t1)]
        for col in ['White Probe Temperature', 'Black Probe Temperature']:
            if col in df.columns and d_idx < len(depth_list):
                v = df[col].dropna()
                if len(v) > 100:
                    # Use 2nd–98th percentile range to exclude short-lived spikes
                    # (e.g. refreezing meltwater events in Chessjen)
                    amp = float(np.percentile(v, 98) - np.percentile(v, 2))
                    pairs.append((depth_list[d_idx], amp))
                d_idx += 1
    return sorted(pairs, key=lambda x: x[0])

def _fit_zaa(pairs):
    z = np.array([p[0] for p in pairs])
    a = np.array([p[1] for p in pairs])
    deepest_amp = a[np.argmax(z)]
    if deepest_amp >= _AMP_THRESHOLD or len(pairs) < 3:
        return None, deepest_amp
    try:
        popt, _ = curve_fit(lambda z, A0, d: A0 * np.exp(-z / d),
                            z, a, p0=[a[0], 10.0], maxfev=5000)
        A0, d = popt
        if A0 <= 0 or d <= 0:
            return None, deepest_amp
        zaa = -d * np.log(_ZAA_SIGNAL / A0)
        return round(zaa, 1), deepest_amp
    except Exception:
        return None, deepest_amp

zaa_depths = {}
for _code in ["AH", "CJ", "HS", "SR", "GT", "CV"]:
    _pairs = _sensor_amplitudes(_code)
    _zaa, _deep_amp = _fit_zaa(_pairs)
    zaa_depths[_code] = _zaa
    _status = f"ZAA ≈ {_zaa} m" if _zaa else f"skipped (deepest amp = {_deep_amp:.3f}°C)"
    print(f"{_code}: {_status}  |  sensor depths/amps: {[(round(p[0],1), round(p[1],3)) for p in _pairs]}")
```

```python
# ── Seasonal amplitude and temperature statistics — all six sites ─────────────
# Computes per-sensor statistics then summarises across all sites.
# Requires: _files, _depths, labels defined in the ZAA computation cell above.

from src.thermistor_processing import ThermistorData

_EQUIL_DAYS = 14   # days to skip at start (freeze-up)

_glacier_names = {
    "AH": "Alphubel South",
    "CJ": "Chessjengletscher",
    "HS": "Hohsaasgletscher",
    "SR": "Glacier du Sex Rouge",
    "GT": "Glacier de Tortin",
    "CV": "Vadret dal Corvatsch",
}

rows = []
for _code in ["AH", "CJ", "HS", "SR", "GT", "CV"]:
    depth_list = _depths[_code]
    d_idx = 0
    for fpath in _files[_code]:
        td = ThermistorData(fpath, delimiter=",")
        df = td.get_ntc_data().sort_values("TIME")
        t0, t1 = df["TIME"].min(), df["TIME"].max()
        t0 = t0 + pd.Timedelta(days=_EQUIL_DAYS)
        df = df[(df["TIME"] >= t0) & (df["TIME"] <= t1)]

        for col, probe in [("White Probe Temperature", "White"),
                           ("Black Probe Temperature", "Black")]:
            if col not in df.columns or d_idx >= len(depth_list):
                d_idx += 1
                continue
            v = df[col].dropna()
            if len(v) < 100:
                d_idx += 1
                continue

            amp_robust = float(np.percentile(v, 98) - np.percentile(v, 2))
            t_mean     = float(v.mean())
            t_min      = float(v.min())
            t_max      = float(v.max())

            # Timing of seasonal minimum (smoothed monthly)
            _monthly = df.set_index("TIME")[col].resample("ME").mean().dropna()
            t_min_month = _monthly.idxmin().strftime("%b") if not _monthly.empty else "—"
            t_max_month = _monthly.idxmax().strftime("%b") if not _monthly.empty else "—"

            rows.append({
                "Glacier":        _glacier_names[_code],
                "Code":           _code,
                "Sensor":         f"{_code}{len([r for r in rows if r['Code']==_code])//2 + 1}TT/{probe[0]}",
                "Depth (m)":      depth_list[d_idx],
                "Mean T (°C)":    round(t_mean, 2),
                "Min T (°C)":     round(t_min,  2),
                "Max T (°C)":     round(t_max,  2),
                "Amplitude (°C)": round(amp_robust, 3),
                "T-min month":    t_min_month,
                "T-max month":    t_max_month,
            })
            d_idx += 1

df_stats = pd.DataFrame(rows)

# ── Per-site summary ──────────────────────────────────────────────────────────
print("=" * 70)
print("PER-SENSOR STATISTICS")
print("=" * 70)
print(df_stats.to_string(index=False))

print()
print("=" * 70)
print("OVERALL SUMMARY (all sensors combined)")
print("=" * 70)
print(f"  N sensors:               {len(df_stats)}")
print(f"  Mean temperature:        {df_stats['Mean T (°C)'].mean():.2f} °C  "
      f"(range {df_stats['Mean T (°C)'].min():.2f} to {df_stats['Mean T (°C)'].max():.2f} °C)")
print(f"  Seasonal amplitude:      {df_stats['Amplitude (°C)'].mean():.3f} °C  "
      f"(range {df_stats['Amplitude (°C)'].min():.3f} to {df_stats['Amplitude (°C)'].max():.3f} °C)")
print(f"  Coldest sensor:          {df_stats.loc[df_stats['Mean T (°C)'].idxmin(), 'Glacier']}  "
      f"({df_stats['Mean T (°C)'].min():.2f} °C)")
print(f"  Warmest sensor:          {df_stats.loc[df_stats['Mean T (°C)'].idxmax(), 'Glacier']}  "
      f"({df_stats['Mean T (°C)'].max():.2f} °C)")
print(f"  Largest amplitude:       {df_stats.loc[df_stats['Amplitude (°C)'].idxmax(), 'Glacier']}  "
      f"({df_stats['Amplitude (°C)'].max():.3f} °C)")
print(f"  Smallest amplitude:      {df_stats.loc[df_stats['Amplitude (°C)'].idxmin(), 'Glacier']}  "
      f"({df_stats['Amplitude (°C)'].min():.3f} °C)")

print()
print("=" * 70)
print("PER-SITE AMPLITUDE SUMMARY")
print("=" * 70)
_site_amp = df_stats.groupby("Glacier")["Amplitude (°C)"].agg(["min","max","mean"]).round(3)
print(_site_amp.to_string())
```

### Alternative: depth-colored version (magma colormap, linestyle = borehole)

```python
# ── Helpers ───────────────────────────────────────────────────────────────────
def _smooth_df_dc(df, days):
    if df is None or df.empty or not days:
        return df
    d = df.copy()
    d["TIME"] = pd.to_datetime(d["TIME"])
    d = d.sort_values("TIME").drop_duplicates("TIME").set_index("TIME")
    d = d.resample("6H").mean().interpolate("time", limit_direction="both")
    win = f"{float(days)}D"
    for col in ("White Probe Temperature", "Black Probe Temperature"):
        if col in d:
            d[col] = d[col].rolling(win, min_periods=1, center=True).mean()
    return d.reset_index()

def _first_nan_dc(series):
    arr = np.array(series)
    mask = np.isnan(arr)
    return int(np.argmax(mask)) if mask.any() else len(arr)

def _load_bh_data_dc(plotter, smooth):
    fps   = getattr(plotter, "file_paths", [getattr(plotter, "file_path", None)])
    delim = getattr(plotter, "delimiter", ",")
    dfs = []
    for fp in fps:
        if fp:
            df = ThermistorData(fp, delim, None).get_ntc_data()
            if smooth and float(smooth) > 0:
                df = _smooth_df_dc(df, smooth)
            df["TIME"] = pd.to_datetime(df["TIME"])
            df.sort_values("TIME", inplace=True)
        else:
            df = pd.DataFrame()
        dfs.append(df)
    return dfs

# ── Config ────────────────────────────────────────────────────────────────────
_top    = ("AH", "CJ", "HS")
_bot    = ("SR", "GT", "CV")
_excl   = {"GT": ["GT2TT"]}
_smooth = 3
_ly     = -5.2
_bfs    = 18
_dpi    = 300
_plbls  = {"AH": "a", "CJ": "b", "HS": "c", "SR": "d", "GT": "e", "CV": "f"}
_glacier_names = {
    "AH": "Alphubel", "CJ": "Chessjen", "HS": "Hohsaas",
    "SR": "Sex Rouge", "GT": "Tortin",  "CV": "Corvatsch",
}
_ann_fs    = _bfs - 5
_equil_days = 14   # days after deployment to start the x-axis

# Deployment dates
_deploy = {
    "AH": ["2024-08-21"],
    "CJ": ["2024-08-09"],
    "HS": ["2024-08-08"],
    "SR": ["2024-08-06"],
    "GT": ["2024-08-07"],
    "CV": ["2024-08-28"],
}
_deploy_label = {"AH", "SR"}

# ── Colormap (initial sensor depth) ───────────────────────────────────────────
_all_d = []
for _code in _top + _bot:
    _di = depths_initial[_code]
    for _bi, _lab in enumerate(labels[_code]):
        if _excl.get(_code) and _lab in _excl[_code]:
            continue
        _all_d += [_di[_bi * 2], _di[_bi * 2 + 1]]

_all_d = [d for d in _all_d if d is not None]
_dmin, _dmax = min(_all_d), max(_all_d)
_cmap_full = mcm.get_cmap("magma_r")
_cmap = mcolors.LinearSegmentedColormap.from_list(
    "magma_r_trunc", _cmap_full(np.linspace(0.15, 1.0, 256))
)
_norm = mcolors.Normalize(vmin=_dmin, vmax=_dmax)

def _dc_color(d):
    return _cmap(_norm(d))

# ── Build figure ──────────────────────────────────────────────────────────────
_fig = plt.figure(figsize=(15.5, 11.5), dpi=_dpi)
_gs  = GridSpec(3, 3, figure=_fig,
                height_ratios=[0.08, 1, 1],
                wspace=0.05, hspace=0.20,
                left=0.07, right=0.98, top=0.91, bottom=0.06)

_lw = 3.2
for _ri, _row in enumerate([_top, _bot]):
    _shared_y = None
    for _ci, _code in enumerate(_row):
        _ax = _fig.add_subplot(_gs[_ri + 1, _ci], sharey=_shared_y)
        if _shared_y is None:
            _shared_y = _ax

        _dfs   = _load_bh_data_dc(plotters[_code], _smooth)
        _dcur  = list(depths_current[_code])
        _dinit = list(depths_initial[_code])

        # Collect per-sensor depth entries for the in-panel legend
        _leg_entries = []  # (ls, color, init_depth, curr_depth, is_dotted)

        for _bi, _lab in enumerate(labels[_code]):
            if _excl.get(_code) and _lab in _excl[_code]:
                continue

            # Skip failed 9.3 m sensor for Corvatsch
            if _code == "CV" and abs(depths_initial[_code][_bi*2] - 9.3) < 0.05:
                continue
            is_dotted = False
            _ls = "-" if _lab.endswith("1TT") else "--"

            _df = _dfs[_bi] if _bi < len(_dfs) else pd.DataFrame()
            if _df.empty:
                continue

            if _code == "GT" and _lab == "GT1TT":
                _wd_curr, _bd_curr = 1.4, 6.4
            else:
                _wd_curr = _dcur[_bi * 2]
                _bd_curr = _dcur[_bi * 2 + 1]

            _wd_init = _dinit[_bi * 2]
            _bd_init = _dinit[_bi * 2 + 1]

            _fi_w = _first_nan_dc(_df["White Probe Temperature"])
            _fi_b = _first_nan_dc(_df["Black Probe Temperature"])
            _skip_bh2_black = False
            if _lab == "CV2TT":
                _fi_b, _fi_w = 1250, 1250
                _skip_bh2_black = True
            if _lab == "GT2TT":
                _fi_b, _fi_w = 1300, 1250

            _t  = _df["TIME"].values
            _wp = _df["White Probe Temperature"].values
            _bp = _df["Black Probe Temperature"].values

            # White probe
            _ax.plot(_t[:_fi_w], _wp[:_fi_w], ls=_ls, color=_dc_color(_wd_init), lw=_lw)
            if _fi_w < len(_t):
                _ax.plot(_t[_fi_w:], _wp[_fi_w:], ls=":", color=_dc_color(_wd_init), lw=_lw)
            _leg_entries.append((_ls, _dc_color(_wd_init), _wd_init, _wd_curr, is_dotted))

            # Black probe
            if _bd_init is not None and not _skip_bh2_black:
              _ax.plot(_t[:_fi_b], _bp[:_fi_b], ls=_ls, color=_dc_color(_bd_init), lw=_lw)
              if _fi_b < len(_t):
                _ax.plot(_t[_fi_b:], _bp[_fi_b:], ls=":", color=_dc_color(_bd_init), lw=_lw)
              _leg_entries.append((_ls, _dc_color(_bd_init), _bd_init, _bd_curr, is_dotted))

            # Chessjen 8.5m 2TT annotation at a fixed date
            if _code == "CJ":
                for _bi, _lab in enumerate(labels[_code]):
                    # Check for 2TT and initial depth close to 8.5
                    if _lab.endswith("2TT") and np.isclose(_dinit[_bi * 2], 8.5, atol=0.3):
                        _df = _dfs[_bi]
                        if not _df.empty:
                            target_date = pd.Timestamp("2025-07-20")
                            idx = np.argmin(np.abs(_df["TIME"] - target_date))
                            ann_time = _df["TIME"].values[idx]
                            ann_temp = _df["White Probe Temperature"].values[idx]
                            _ax.annotate(
                                "refreezing meltwater",
                                xy=(ann_time, ann_temp),
                                xytext=(ann_time + np.timedelta64(-5, 'D'), ann_temp + 0.5),
                                arrowprops=dict(arrowstyle="->", color="gray", lw=1.5),
                                fontsize=_bfs - 4, color="gray", ha="right", va="top"
                            )
                        break

        # Deployment lines (no annotation)
        for _dd in _deploy.get(_code, []):
            _ddt = pd.to_datetime(_dd)
            _ax.axvline(_ddt, color="gray", lw=1.5, ls="solid", alpha=0.8, zorder=0)

        _ax.axhline(0, color="k", lw=0.8, ls="--", alpha=0.5)
        _ax.set_ylim(_ly, 0.5)
        _loc = mdates.AutoDateLocator(minticks=8, maxticks=11)
        _ax.xaxis.set_major_locator(_loc)
        _ax.xaxis.set_major_formatter(mdates.ConciseDateFormatter(_loc))
        _ax.tick_params(axis="both", labelsize=_bfs - 4)
        _ax.grid(True, which="major", color="lightgrey", linewidth=0.6, zorder=0)

        # Skip equilibration phase
        if _deploy.get(_code):
            _xlim_start = pd.to_datetime(_deploy[_code][0]) + pd.Timedelta(days=_equil_days)
            _ax.set_xlim(left=_xlim_start)

        if _ci > 0:
            _ax.set_ylabel("")
            _ax.tick_params(labelleft=False)
        else:
            _ax.set_ylabel("Temperature [°C]", fontsize=_bfs - 2)

        _gname = f"{_glacier_names[_code]} ({_code})"
        _ax.text(0.02, 0.02, _gname, transform=_ax.transAxes,
                 ha="left", va="bottom", fontsize=_bfs - 3, weight="bold",
                 bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="k", alpha=0.7))
        _ax.text(0.02, 0.98, f"({_plbls.get(_code, '')})", transform=_ax.transAxes,
                 ha="left", va="top", fontsize=_bfs, weight="bold",
                 bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=1.0))

        # ── In-panel depth legend ──
        if _leg_entries:
            _handles, _lbls = [], []
            for _ls, _col, _di, _dc, _is_dotted in _leg_entries:
                _handles.append(Line2D([0], [0], color=_col, lw=_lw, ls=_ls))
                _lbls.append(f"{_di:.1f}→{_dc:.1f} m")
            # Add Z_init and Z_final header above legend, spaced and with arrow
            legend_loc = "lower right"
            if _code == "CV":
                legend_loc = "upper center"
            leg = _ax.legend(_handles, _lbls,
                       loc=legend_loc,
                       fontsize=_ann_fs,
                       frameon=True, fancybox=False,
                       edgecolor="black", framealpha=0.92, facecolor="white",
                       handlelength=1.6, handleheight=0.8,
                       borderpad=0.4, labelspacing=0.25,
                       title="$Z_{init}\\;\\rightarrow\\;Z_{final}$",
                       title_fontsize=_ann_fs)

# ── Legend bar ────────────────────────────────────────────────────────────────
_frame_l, _frame_b, _frame_w, _frame_h = 0.07, 0.862, 0.91, 0.124
_frame_ax = _fig.add_axes([_frame_l, _frame_b, _frame_w, _frame_h])
_frame_ax.set_xlim(0, 1); _frame_ax.set_ylim(0, 1)
_frame_ax.set_axis_on()
_frame_ax.patch.set_facecolor("white")
_frame_ax.tick_params(left=False, bottom=False, labelleft=False, labelbottom=False)
for spine in _frame_ax.spines.values():
    spine.set_visible(True); spine.set_linewidth(1.0); spine.set_color("black")
_frame_ax.text(0.007, 0.93, "Legend", transform=_frame_ax.transAxes,
               ha="left", va="top", fontsize=_bfs - 3, fontweight="bold")

_cbar_ax = _fig.add_axes([_frame_l + 0.03, _frame_b + 0.058, 0.51, 0.020])
_cb = ColorbarBase(_cbar_ax, cmap=_cmap, norm=_norm, orientation="horizontal")
_cb.set_label("Initial sensor depth [m]", fontsize=_bfs - 4, labelpad=2)
_cb.ax.tick_params(labelsize=_bfs - 3)

# Restore 1TT/2TT and Borehole header in main legend bar
_leg_ax = _fig.add_axes([_frame_l + 0.60, _frame_b - 0.002, 0.30, _frame_h + 0.004])
_leg_ax.set_axis_off()
_leg_handles = [
    Line2D([0], [0], color="k", lw=_lw, ls="-",  label="1TT"),
    Line2D([0], [0], color="k", lw=_lw, ls="--", label="2TT"),
]
_leg_ax.legend(handles=_leg_handles, loc="center", ncol=2,
               frameon=False, fontsize=_bfs - 2,
               title="Borehole", title_fontsize=_bfs - 4)

# ── ZAA annotations in panel depth legends ────────────────────────────────────
# zaa_depths is computed in the ZAA computation cell above.
# Only annotate panels where ZAA was successfully estimated.
_panel_order = list(_top) + list(_bot)
_data_axes = [ax for ax in _fig.get_axes()
              if ax.get_subplotspec() is not None
              and ax.get_subplotspec().get_gridspec() is _gs
              and ax.get_subplotspec().rowspan.start > 0]
_data_axes.sort(key=lambda ax: (ax.get_subplotspec().rowspan.start,
                                ax.get_subplotspec().colspan.start))
for _ax, _code in zip(_data_axes, _panel_order):
    _zaa = zaa_depths.get(_code) if 'zaa_depths' in dir() else None
    if _zaa is not None:
        _ax.text(0.02, 0.09, f"ZAA ≈ {_zaa:.0f} m",
                 transform=_ax.transAxes, ha="left", va="bottom",
                 fontsize=_ann_fs, color="black", style="italic",
                 bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.85))

_fig.supxlabel("Time", fontsize=_bfs, y=0.01)
_fig.savefig(
    os.path.join(project_root, "figures", "paper", "fig03_tt_timeseries_all6_depth_colored.pdf"),
    dpi=_dpi, bbox_inches="tight"
)
plt.show()
```
