# Study Sites, Maps, and Ice Thickness

This notebook produces the field-site overview figures and GPR-derived ice thickness maps used in the paper. It is divided into two main sections:

**Part 1 – Overview map and field site photos**
- Switzerland-scale map with the six field sites (Alphubel, Chessjen, Hohsaas, Sex Rouge, Tortin, Corvatsch)
- Photo mosaic combining satellite overview with ground-level glacier photographs

**Part 2 – GPR ice thickness maps and site geometry**
- GPR bedrock picks loaded and interpolated to a regular grid for each glacier
- Swisstopo orthophotos downloaded and used as basemaps
- Borehole locations overlaid on each thickness map
- Geometric glacier properties (area, mean slope, elevation range) computed from SGI outlines and DEM tiles

### Individual glacier photographs

```python
# Glacier codes and coordinates (EPSG:2056)
field_sites = {
    "AH": (2633182.66, 1100625.66),
    "CJ": (2636979.60, 1101745.93),
    "HS": (2643539.74, 1109801.47),
    "SR": (2645000.00, 1105000.00),
    "GT": (2590000.00, 1103000.00),
    "CV": (2783145.24, 1143198.41),
}

glacier_image_dir = os.path.join(project_root, "products", "figures", "maps", "glacier_images")
output_dir = os.path.join(project_root, "products", "figures", "maps")

def _find_image(code: str) -> Path | None:
    d = Path(glacier_image_dir)
    for p in d.glob(f"{code}*lowres.jpg"):
        return p
    return None

def _crop_center(im, target_size):
    w, h = im.size
    tw, th = target_size
    left = (w - tw) // 2
    top = (h - th) // 2
    right = left + tw
    bottom = top + th
    return im.crop((left, top, right, bottom))

def _load_img(code, target_size=None):
    p = _find_image(code)
    if p is None:
        return None
    im = Image.open(p).convert("RGBA")
    if target_size:
        im = im.resize(target_size, Image.LANCZOS)
    return np.array(im)

# Save each glacier image individually, cropped to same format
panel_labels = {"AH": "a", "CJ": "b", "HS": "c", "SR": "d", "GT": "e", "CV": "f"}
color_map = {"AH": "k", "CJ": "k", "HS": "k", "SR": "k", "GT": "k", "CV": "k"}
ABBR_FONTSIZE = 18
TARGET_SIZE = (950, 717)  # Change as needed

for code in field_sites:
    arr = _load_img(code, target_size=TARGET_SIZE)
    if arr is not None:
        fig, ax = plt.subplots(figsize=(6,6), dpi=300)
        ax.imshow(arr)
        ax.set_xticks([]); ax.set_yticks([]); ax.set_axis_off()
        ax.text(0.02, 0.02, code, transform=ax.transAxes, ha="left", va="bottom",
                fontsize=ABBR_FONTSIZE, fontweight="bold", color=color_map[code],
                fontname="Arial",
                bbox=dict(boxstyle="round,pad=0.15", fc="white", ec=color_map[code], alpha=0.7))
        ax.text(0.02, 0.98, f"({panel_labels[code]})", transform=ax.transAxes, ha='left', va='top',
            fontsize=ABBR_FONTSIZE, fontweight='bold',
            color='black', zorder=21,
            bbox=dict(boxstyle='round,pad=0.15', fc='white', ec='none', alpha=1.0))
        fig.savefig(f"{output_dir}/{code}_glacier_image.png", dpi=300, bbox_inches="tight", pad_inches=0.05)
        plt.close(fig)
```

### Switzerland location inset

```python
def add_switzerland_inset(
    ax,
    overview_extent,
    inset_position=(0.72, -0.02, 0.26, 0.26),
    buffer_km=40,
    shp_path=os.path.join(project_root, "products", "figures", "maps", "World_Countries_(Generalized)_-573431906301700955", "World_Countries_Generalized.shp"),
):
    """
    Add a small inset map of Switzerland showing the location of the overview extent,
    using the locally stored World Countries (Generalized) shapefile.
    """
    from matplotlib.patches import Rectangle
    from shapely.geometry import box
    import geopandas as gpd

    world = gpd.read_file(shp_path)
    # Find ISO 2-letter field
    iso_col = next((c for c in ["ISO2", "ISO_A2", "ISO"] if c in world.columns), None)
    if iso_col is None:
        raise RuntimeError("ISO country code column not found in shapefile.")

    world = world.to_crs(epsg=2056)

    ax_inset = ax.inset_axes(inset_position)

    # Add "CH" label
    ax_inset.patch.set_facecolor("white")
    ax_inset.patch.set_alpha(0.9)
    ax_inset.text(0.5, 0.5, "CH", transform=ax_inset.transAxes,
                  ha="center", va="center", fontsize=14, fontweight="bold",
                  color="black", zorder=4)

    ax_inset.patch.set_facecolor("white")
    ax_inset.patch.set_alpha(0.9)

    # Switzerland extent and buffer
    ch_extent = (2480000, 2840000, 1070000, 1300000)
    buf = buffer_km * 1000
    bbox_geom = box(ch_extent[0] - buf, ch_extent[2] - buf,
                    ch_extent[1] + buf, ch_extent[3] + buf)
    bbox_poly = gpd.GeoSeries([bbox_geom], crs="EPSG:2056")

    neighbors = gpd.clip(world, bbox_poly.iloc[0])
    ch = neighbors[neighbors[iso_col] == "CH"]

    if not neighbors.empty:
        neighbors.plot(ax=ax_inset, facecolor="#e8e8e8", edgecolor="gray", linewidth=0.6, zorder=1)
    if not ch.empty:
        ch.plot(ax=ax_inset, facecolor="lightgray", edgecolor="black", linewidth=0.9, zorder=2)

    # Red rectangle for the provided overview_extent
    rect = Rectangle(
        (overview_extent[0], overview_extent[2]),
        overview_extent[1] - overview_extent[0],
        overview_extent[3] - overview_extent[2],
        linewidth=1.8, edgecolor="red", facecolor="none", zorder=5,
    )
    ax_inset.add_patch(rect)

    ax_inset.set_xlim(ch_extent[0] - buf, ch_extent[1] + buf)
    ax_inset.set_ylim(ch_extent[2] - buf, ch_extent[3] + buf)
    ax_inset.set_aspect("equal")
    ax_inset.tick_params(left=False, bottom=False, labelleft=False, labelbottom=False)
    for spine in ax_inset.spines.values():
        spine.set_edgecolor("black"); spine.set_linewidth(1)
    return ax_inset
```

```python
def merge_gpr_line_data(coords_file, surface_file, bed_file, line_name, output_file):
    """
    Merge GPR line coordinates with surface and bed elevation data.
    Creates format compatible with load_points_from_csv with thickness pre-calculated.
    
    Parameters
    ----------
    coords_file : str
        Path to CSV file with columns: begin, end, begin_2, end_2, xcoord, ycoord
    surface_file : str
        Path to text file with surface elevations (one per line)
    bed_file : str
        Path to text file with bed elevations (one per line)
    line_name : str
        Name of the profile/line (e.g., "centerline1", "centerline2")
    output_file : str
        Path where the merged CSV should be saved
    
    Returns
    -------
    df : pd.DataFrame
        Merged dataframe with all information
    """
    
    # Load coordinates
    coords_df = pd.read_csv(coords_file)
    
    # Load surface elevations (remove quotes if present)
    with open(surface_file, 'r') as f:
        surface_elevs = [float(line.strip().strip('"')) for line in f if line.strip()]
    
    # Load bed elevations
    with open(bed_file, 'r') as f:
        bed_elevs = [float(line.strip().strip('"')) for line in f if line.strip()]
    
    # Check that all arrays have same length
    n_coords = len(coords_df)
    n_surface = len(surface_elevs)
    n_bed = len(bed_elevs)
    
    print(f"Line {line_name}:")
    print(f"  Coordinates: {n_coords} points")
    print(f"  Surface: {n_surface} values")
    print(f"  Bed: {n_bed} values")
    
    # Use the minimum length to avoid indexing errors
    n_points = min(n_coords, n_surface, n_bed)
    
    if n_coords != n_surface or n_coords != n_bed:
        print(f"  WARNING: Lengths don't match! Using first {n_points} points.")
    
    # Create ONE row per point with thickness and bed elevation filled
    results = []
    for i in range(n_points):
        surface_elev = surface_elevs[i]
        bed_elev = bed_elevs[i]
        thickness = surface_elev - bed_elev  # Calculate ice thickness
        
        results.append({
            'Profile': line_name,
            'Channel': '',
            'Interpretation': 'Horizon interpretation',  # Match target format
            'Trace number': i + 1,
            'X': coords_df.loc[i, 'xcoord'],
            'Y': coords_df.loc[i, 'ycoord'],
            'Depth': thickness,           # Ice thickness (surface - bed)
            'Travel time': '',
            'Elevation': bed_elev,        # Bed elevation (NOT surface!)
            'Amplitude': ''
        })
    
    # Create DataFrame
    df = pd.DataFrame(results)
    
    # Save to CSV
    df.to_csv(output_file, index=False)
    print(f"  Saved {len(df)} points to: {output_file}\n")
    
    return df


def merge_multiple_lines(line_configs, output_combined):
    """
    Merge multiple GPR lines into a single file.
    
    Parameters
    ----------
    line_configs : list of dict
        Each dict should have keys: 'coords_file', 'surface_file', 'bed_file', 'line_name'
    output_combined : str
        Path where the combined CSV should be saved
    
    Returns
    -------
    df_all : pd.DataFrame
        Combined dataframe with all lines
    """
    
    all_dfs = []
    
    for config in line_configs:
        # Process each line individually
        temp_output = config.get('temp_output', None)
        if temp_output is None:
            temp_output = f"/tmp/{config['line_name']}_temp.csv"
        
        df = merge_gpr_line_data(
            coords_file=config['coords_file'],
            surface_file=config['surface_file'],
            bed_file=config['bed_file'],
            line_name=config['line_name'],
            output_file=temp_output
        )
        
        all_dfs.append(df)
    
    # Combine all lines
    df_all = pd.concat(all_dfs, ignore_index=True)
    
    # Save combined file
    df_all.to_csv(output_combined, index=False)
    print(f"Combined file saved to: {output_combined}")
    print(f"Total points: {len(df_all)} ({len(line_configs)} lines)")
    
    return df_all


# Example usage for Hohsaas glacier with 2 lines
line_configs = [
    {
        'line_name': 'centerline1',
        'coords_file': gpr_data_dir + '20250928_Hohsaas_data_drone_gpr/flights_and_profiles/centerline_1_three_bhs.csv',
        'surface_file': gpr_data_dir + '20250928_Hohsaas_data_drone_gpr/centerline/centerline_1_glacier_surface.txt',
        'bed_file': gpr_data_dir + '20250928_Hohsaas_data_drone_gpr/centerline/centerline_1_bed.txt',
        'temp_output': gpr_output_dir + '20250928_Hohsaas_centerline_1_picks.csv'
    },
    {
        'line_name': 'centerline2',
        'coords_file': gpr_data_dir + '20250928_Hohsaas_data_drone_gpr/flights_and_profiles/centerline_2_three_bhs.csv',
        'surface_file': gpr_data_dir + '20250928_Hohsaas_data_drone_gpr/centerline/centerline_2_glacier_surface.txt',
        'bed_file': gpr_data_dir + '20250928_Hohsaas_data_drone_gpr/centerline/centerline_2_bed.txt',
        'temp_output': gpr_output_dir + '20250928_Hohsaas_centerline_2_picks.csv'
    }
]

# Process all lines and create combined file
df_combined = merge_multiple_lines(
    line_configs=line_configs,
    output_combined=gpr_output_dir + '20250928_Hohsaas_drone_gpr_centerline_picks.csv'
)

# Display sample of the result
print("\nSample of combined data:")
print(df_combined.head(20))

# Load combined GPR line data
drone_gpr_lines_hohsaas = gpr.load_points_from_csv(gpr_output_dir + '20250928_Hohsaas_drone_gpr_centerline_picks.csv', epsg=2056, source_epsg=2056, drop_duplicates=True, aggregate_duplicates='mean')
```

```python
# Compute glacier geometric properties from DEM tiles and SGI outlines



def parse_xyzn_polygons(xyzn_path, jump_threshold=200.0):
    """
    Parse a .xyzn file that may contain multiple concatenated glacier outlines.
    Splits into separate Polygon objects wherever consecutive points are
    more than `jump_threshold` metres apart.

    Returns
    -------
    list of shapely.geometry.Polygon, sorted largest → smallest by area.
    """
    coords = []
    with open(xyzn_path) as fh:
        for line in fh:
            parts = line.strip().split()
            if len(parts) < 2:
                continue
            try:
                coords.append((float(parts[0]), float(parts[1])))
            except ValueError:
                continue

    if not coords:
        return []

    # Split at large jumps
    segments = []
    current = [coords[0]]
    for i in range(1, len(coords)):
        dx = coords[i][0] - coords[i - 1][0]
        dy = coords[i][1] - coords[i - 1][1]
        if (dx**2 + dy**2) ** 0.5 > jump_threshold:
            segments.append(current)
            current = []
        current.append(coords[i])
    segments.append(current)

    polygons = []
    for seg in segments:
        if len(seg) >= 3:
            try:
                polygons.append(Polygon(seg))
            except Exception:
                pass

    return sorted(polygons, key=lambda p: p.area, reverse=True)


def geometric_properties_from_dem(xyzn_path, dem_tile_paths,
                                   reference_point=None, polygon_index=None):
    """
    Compute glacier geometric properties from DEM clipped to SGI outline.

    If the xyzn file contains multiple polygons (concatenated outlines),
    the correct one is selected by:
      1. `reference_point` (x, y) in LV95 – picks the polygon that contains it.
      2. `polygon_index` (int) – picks by index (0 = largest).
      3. Default: largest polygon.

    Parameters
    ----------
    xyzn_path       : str   – path to .xyzn file (LV95 / EPSG:2056)
    dem_tile_paths  : list  – SwissALTI3D GeoTIFF tile paths
    reference_point : tuple – (x, y) in LV95 to select the right polygon
    polygon_index   : int   – fallback polygon selector (0 = largest)

    Returns
    -------
    dict with: area_km2, min_elev, max_elev, elev_range, mean_elev,
               mean_slope_deg, mean_aspect_deg, aspect_cardinal, n_polygons
    """
    polygons = parse_xyzn_polygons(xyzn_path)
    if not polygons:
        raise ValueError(f"No valid polygons found in {xyzn_path}")

    # Select polygon
    if reference_point is not None:
        pt = Point(reference_point)
        matches = [p for p in polygons if p.contains(pt)]
        if not matches:
            raise ValueError(
                f"No polygon in {xyzn_path} contains reference point {reference_point}. "
                f"Found {len(polygons)} polygon(s) with areas: "
                f"{[round(p.area/1e6, 3) for p in polygons]} km²"
            )
        polygon = matches[0]
    elif polygon_index is not None:
        polygon = polygons[polygon_index]
    else:
        polygon = polygons[0]  # largest by default

    area_km2 = polygon.area / 1e6

    # --- Mosaic DEM tiles ---
    srcs = [rasterio.open(p) for p in dem_tile_paths]
    mosaic, transform = rio_merge(srcs)
    nodata = srcs[0].nodata
    crs = srcs[0].crs
    for s in srcs:
        s.close()

    dem_full = mosaic[0].astype(np.float64)
    dem_full[dem_full == nodata] = np.nan

    # --- Clip DEM to selected polygon ---
    with MemoryFile() as memfile:
        with memfile.open(
            driver='GTiff', height=dem_full.shape[0], width=dem_full.shape[1],
            count=1, dtype='float64', crs=crs, transform=transform, nodata=np.nan
        ) as ds:
            ds.write(dem_full, 1)
        with memfile.open() as ds:
            clipped, _ = rio_mask(ds, [mapping(polygon)], crop=False, nodata=np.nan)

    dem = clipped[0]
    valid = np.isfinite(dem)

    if not valid.any():
        raise ValueError(f"No valid DEM pixels within selected polygon in {xyzn_path}")

    # --- Elevation statistics ---
    elevs = dem[valid]

    # --- Slope and aspect from DEM gradient ---
    # SwissALTI3D: 0.5 m pixels, rows run N→S (transform.e < 0)
    res_x = abs(transform.a)
    res_y = abs(transform.e)

    dz_drow, dz_dcol = np.gradient(dem, res_y, res_x)
    dz_dN = -dz_drow   # North = decreasing row index
    dz_dE =  dz_dcol   # East  = increasing col index

    slope_deg = np.degrees(np.arctan(np.sqrt(dz_dN**2 + dz_dE**2)))
    aspect_deg_full = np.degrees(np.arctan2(-dz_dE, -dz_dN)) % 360

    mean_slope = float(np.nanmean(slope_deg[valid]))

    a_rad = np.radians(aspect_deg_full[valid])
    mean_aspect = float(
        np.degrees(np.arctan2(np.nanmean(np.sin(a_rad)),
                               np.nanmean(np.cos(a_rad)))) % 360
    )
    sectors = ["N", "NE", "E", "SE", "S", "SW", "W", "NW"]
    aspect_cardinal = sectors[int((mean_aspect + 22.5) / 45) % 8]

    return dict(
        area_km2        = round(area_km2, 3),
        min_elev        = round(float(np.nanmin(elevs))),
        max_elev        = round(float(np.nanmax(elevs))),
        elev_range      = round(float(np.nanmax(elevs) - np.nanmin(elevs))),
        mean_elev       = round(float(np.nanmean(elevs))),
        mean_slope_deg  = round(mean_slope, 1),
        mean_aspect_deg = round(mean_aspect, 1),
        aspect_cardinal = aspect_cardinal,
        n_polygons      = len(polygons),
    )


def mean_slope_for_polygon(polygon, dem_tile_paths):
    """Return mean slope (degrees) of DEM pixels within `polygon`."""
    srcs = [rasterio.open(p) for p in dem_tile_paths]
    mosaic, transform = rio_merge(srcs)
    nodata = srcs[0].nodata
    crs = srcs[0].crs
    for s in srcs:
        s.close()

    dem_full = mosaic[0].astype(np.float64)
    dem_full[dem_full == nodata] = np.nan

    with MemoryFile() as memfile:
        with memfile.open(
            driver="GTiff", height=dem_full.shape[0], width=dem_full.shape[1],
            count=1, dtype="float64", crs=crs, transform=transform, nodata=np.nan
        ) as ds:
            ds.write(dem_full, 1)
        with memfile.open() as ds:
            clipped, _ = rio_mask(ds, [mapping(polygon)], crop=False, nodata=np.nan)

    dem = clipped[0]
    valid = np.isfinite(dem)
    if not valid.any():
        return float("nan")

    res_x = abs(transform.a)
    res_y = abs(transform.e)
    dz_drow, dz_dcol = np.gradient(dem, res_y, res_x)
    slope_deg = np.degrees(np.arctan(np.sqrt((-dz_drow)**2 + dz_dcol**2)))
    return round(float(np.nanmean(slope_deg[valid])), 1)


# ---- Run for all six glaciers ----
# For HS: pass a borehole coordinate to select the Hohsaasgletscher polygon
# (the xyzn file also contains a larger neighbouring glacier outline)
HS_REFERENCE = (2643571.23, 1109771.59)  # HS1G borehole, LV95

glacier_inputs = {
    "AH": dict(xyzn=xyzn_path_AH, dems=dem_tiles_AH),
    "CJ": dict(xyzn=xyzn_path_CJ, dems=dem_tiles_CJ),
    "HS": dict(xyzn=xyzn_path_HS, dems=dem_tiles_HS, reference_point=HS_REFERENCE),
    "SR": dict(xyzn=xyzn_path_SR, dems=dem_tiles_SR),
    "GT": dict(xyzn=xyzn_path_TT, dems=dem_tiles_GT),
    "CV": dict(xyzn=xyzn_path_CV, dems=dem_tiles_CV),
}

# Reviewer feedback: averaging slope over the whole glacier outline can be
# unrepresentative of the actual study area (e.g. Tortin's GPR coverage is
# markedly less steep than the glacier average). mean_slope_deg is therefore
# recomputed restricted to the GPR coverage polygon (convex hull of GPR
# points) for each site; the glacier-wide value is kept alongside for reference.
coverage_by_abbr = {
    "AH": coverage_AH.geometry.iloc[0],
    "CJ": coverage_CJ.geometry.iloc[0],
    "HS": coverage_HS.geometry.iloc[0],
    "SR": coverage_SR.geometry.iloc[0],
    "GT": coverage_GT.geometry.iloc[0],
    "CV": coverage_CV.geometry.iloc[0],
}

results = {}
for abbr, kwargs in glacier_inputs.items():
    xyzn = kwargs.pop("xyzn")
    dems = kwargs.pop("dems")
    props = geometric_properties_from_dem(xyzn, dems, **kwargs)
    props['mean_slope_deg_glacier_wide'] = props['mean_slope_deg']
    props['mean_slope_deg'] = mean_slope_for_polygon(coverage_by_abbr[abbr], dems)
    results[abbr] = props
    print(f"{abbr}: {props['n_polygons']} polygon(s) in file, "
          f"selected area={props['area_km2']} km², "
          f"slope (GPR survey area)={props['mean_slope_deg']}°, "
          f"slope (glacier-wide)={props['mean_slope_deg_glacier_wide']}°, "
          f"aspect={props['aspect_cardinal']}")

summary = pd.DataFrame(results).T
summary.index.name = "Glacier"
print("\n", summary.to_string())


# ============================================================
# Additional slope calculations for Chessjen
# ============================================================

# --- Identify Chessjen polygons ---
cj_polygons = parse_xyzn_polygons(xyzn_path_CJ)
print("Chessjen: {} polygon(s)".format(len(cj_polygons)))

# Mosaic CJ DEM once for elevation queries
_srcs_cj = [rasterio.open(p) for p in dem_tiles_CJ]
_mosaic_cj, _transform_cj = rio_merge(_srcs_cj)
_nodata_cj = _srcs_cj[0].nodata
_crs_cj    = _srcs_cj[0].crs
for _s in _srcs_cj: _s.close()
_dem_cj = _mosaic_cj[0].astype(np.float64)
_dem_cj[_dem_cj == _nodata_cj] = np.nan

def _mean_elev_poly(poly):
    with MemoryFile() as mf:
        with mf.open(driver="GTiff", height=_dem_cj.shape[0], width=_dem_cj.shape[1],
                     count=1, dtype="float64", crs=_crs_cj,
                     transform=_transform_cj, nodata=np.nan) as ds:
            ds.write(_dem_cj, 1)
        with mf.open() as ds:
            cl, _ = rio_mask(ds, [mapping(poly)], crop=False, nodata=np.nan)
    vals = cl[0][np.isfinite(cl[0])]
    return float(np.nanmean(vals)) if len(vals) else float("nan")

_cj_mean_elevs = [_mean_elev_poly(p) for p in cj_polygons]
for idx, (poly, elev) in enumerate(zip(cj_polygons, _cj_mean_elevs)):
    s = mean_slope_for_polygon(poly, dem_tiles_CJ)
    print("  Polygon {}: area={:.4f} km2, mean_elev={:.0f} m, mean_slope={}deg".format(
        idx, poly.area / 1e6, elev, s))

# Lower polygon = lowest mean elevation
_lower_idx = int(np.argmin(_cj_mean_elevs))
lower_poly_cj = cj_polygons[_lower_idx]
slope_cj_lower = mean_slope_for_polygon(lower_poly_cj, dem_tiles_CJ)
print("\nChessjen lower polygon (index {}, mean elev {:.0f} m a.s.l., area {:.4f} km2):".format(
    _lower_idx, _cj_mean_elevs[_lower_idx], lower_poly_cj.area / 1e6))
print("  Mean slope: {}deg".format(slope_cj_lower))

# --- Slope in 50 m buffer around CJ1G ---
_cj1g_geom = boreholes_CJ[boreholes_CJ["name"] == "CJ1G"].iloc[0].geometry
cj1g_buffer_50m = _cj1g_geom.buffer(50)
slope_cj1g_50m = mean_slope_for_polygon(cj1g_buffer_50m, dem_tiles_CJ)
print("\nChessjen CJ1G 50 m buffer:")
print("  Mean slope: {}deg".format(slope_cj1g_50m))
```
